```python
# Load CSV data from GitHub and insert into MongoDB collections
import pandas as pd

base_url = "https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/"

customers_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/customers.csv")
orders_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/orders.csv")
deliveries_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/deliveries.csv")
drivers_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/drivers.csv")
vehicles_df = pd.read_csv( "https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/vehicles.csv")
hubs_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/hubs.csv")
incidents_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/incidents.csv")
complaints_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/complaints.csv")
app_events_df = pd.read_csv("https://raw.githubusercontent.com/raniyaarani13-web/north-star-analytics/refs/heads/main/app_events.csv")

print("All datasets loaded successfully!")
print(f"Customers: {len(customers_df)} rows")
print(f"Orders: {len(orders_df)} rows")
print(f"Deliveries: {len(deliveries_df)} rows")
print(f"Drivers: {len(drivers_df)} rows")
print(f"Vehicles: {len(vehicles_df)} rows")
print(f"Hubs: {len(hubs_df)} rows")
print(f"Incidents: {len(incidents_df)} rows")
print(f"Complaints: {len(complaints_df)} rows")
print(f"App Events: {len(app_events_df)} rows")
```

```python
# CRUD Operations - find() and findOne()

# Find one customer
print("FIND ONE CUSTOMER")
print(db.customers.find_one())

# Find all failed deliveries
print("\nFAILED DELIVERIES (first 5)")
failed = db.deliveries.find({"delivery_status": "Failed"}).limit(5)
for d in failed:
    print(d)

# Count documents by delivery status
print("\nDELIVERY STATUS COUNT")
for status in ["OnTime", "Delayed", "Failed"]:
    count = db.deliveries.count_documents({"delivery_status": status})
    print(f"{status}: {count}")
```

```python
# UPDATE Operations - updateOne() and updateMany()

# Update one customer's account status
print("UPDATE ONE CUSTOMER")
result = db.customers.update_one(
    {"customer_id": "C0001"},
    {"$set": {"account_status": "Premium"}})
print(f"Matched: {result.matched_count}, Modified: {result.modified_count}")

# Verify the update
updated = db.customers.find_one({"customer_id": "C0001"})
print(f"Updated status: {updated['account_status']}")

# Update many - flag all failed deliveries
print("UPDATE MANY - Flag Failed Deliveries")
result2 = db.deliveries.update_many(
    {"delivery_status": "Failed"},
    {"$set": {"requires_review": True}})
print(f"Matched: {result2.matched_count}, Modified: {result2.modified_count}")

# Verify
sample = db.deliveries.find_one({"delivery_status": "Failed"})
print(f"Sample failed delivery requires_review: {sample.get('requires_review')}")
```

```python
# Aggregation 1: Count deliveries by status
print("AGG 1: Delivery Count by Status")
pipeline1 = [
    {"$group": {"_id": "$delivery_status", "count": {"$sum": 1}}},
    {"$sort": {"count": -1}}]
for doc in db.deliveries.aggregate(pipeline1):
    print(doc)

# Aggregation 2: Average rating per hub
import math

# Drop and reinsert deliveries with cleaned data
db.deliveries.drop()

# Clean the dataframe - replace NaN with None (which becomes null in MongoDB)
deliveries_clean = deliveries_df.copy()
deliveries_clean['customer_rating_post_delivery'] = deliveries_clean['customer_rating_post_delivery'].where(
    deliveries_clean['customer_rating_post_delivery'].notna(), None)

# Convert to records and insert
records = []
for record in deliveries_clean.to_dict('records'):
    # Replace any remaining float nan with None
    cleaned = {k: (None if isinstance(v, float) and math.isnan(v) else v)
               for k, v in record.items()}
    records.append(cleaned)

db.deliveries.insert_many(records)
print(f"\nReinserted {db.deliveries.count_documents({})} delivery documents")


print("\nAGG 2: Average Customer Rating per Hub")
pipeline2 = [
    {"$group": {
        "_id": "$hub_id",
        "avg_rating": {"$avg": "$customer_rating_post_delivery"},
        "total_deliveries": {"$sum": 1}}},
    {"$sort": {"avg_rating": -1}}]
for doc in db.deliveries.aggregate(pipeline2):
    print(f"Hub: {doc['_id']}, Avg Rating: {round(doc['avg_rating'],2)}, Total: {doc['total_deliveries']}")

# Aggregation 3: Total complaints by type
print("\nAGG 3: Complaints by Type")
pipeline3 = [
    {"$group": {
        "_id": "$complaint_type",
        "total": {"$sum": 1},
        "avg_resolution_days": {"$avg": "$resolution_days"}}},
    {"$sort": {"total": -1}}]
for doc in db.complaints.aggregate(pipeline3):
    print(f"Type: {doc['_id']}, Total: {doc['total']}, Avg Resolution: {round(doc['avg_resolution_days'],1)} days")

# Aggregation 4: Filter then group - High severity incidents
print("\nAGG 4: High Severity Incidents by Type")
pipeline4 = [
    {"$match": {"severity": "High"}},
    {"$group": {"_id": "$incident_type", "count": {"$sum": 1}}},
    {"$sort": {"count": -1}}]
for doc in db.incidents.aggregate(pipeline4):
    print(doc)

# Aggregation 5: Average order value by service type
print("\nAGG 5: Average Order Value by Service Type")
pipeline5 = [
    {"$group": {
        "_id": "$service_type",
        "avg_value": {"$avg": "$order_value"},
        "total_orders": {"$sum": 1},
        "max_value": {"$max": "$order_value"}}},
    {"$sort": {"avg_value": -1}}]
for doc in db.orders.aggregate(pipeline5):
    print(f"Service: {doc['_id']}, Avg: £{round(doc['avg_value'],2)}, Total: {doc['total_orders']}, Max: £{round(doc['max_value'],2)}")
```

```python
# Aggregation 6: Top 5 drivers by number of deliveries
print("\nAGG 6: Top 5 Drivers by Deliveries")
pipeline6 = [
    {"$group": {
        "_id": "$driver_id",
        "total_deliveries": {"$sum": 1},
        "avg_rating": {"$avg": "$customer_rating_post_delivery"}}},
    {"$sort": {"total_deliveries": -1}},
    {"$limit": 5}]
for doc in db.deliveries.aggregate(pipeline6):
    print(f"Driver: {doc['_id']}, Deliveries: {doc['total_deliveries']}, Avg Rating: {round(doc['avg_rating'],2) if doc['avg_rating'] else 'N/A'}")

# Aggregation 7: Complaints by severity and status
print("\nAGG 7: Complaints by Severity")
pipeline7 = [
    {"$group": {
        "_id": {"severity": "$severity", "status": "$status"},
        "count": {"$sum": 1}}},
    {"$sort": {"count": -1}},
    {"$limit": 8}]
for doc in db.complaints.aggregate(pipeline7):
    print(f"Severity: {doc['_id']['severity']}, Status: {doc['_id']['status']}, Count: {doc['count']}")
```

```python
# Query Optimisation - Measuring baseline performance (No Index)
# We will search for a specific customer_id in the orders collection

print("Performance WITHOUT Index:")
query_filter = {"customer_id": "C0023"}

# Use explain() to see how MongoDB finds the data
execution_stats = db.orders.find(query_filter).explain()['executionStats']

print(f"Execution Time (ms): {execution_stats['executionTimeMillis']}ms")
print(f"Total Docs Examined: {execution_stats['totalDocsExamined']}")
print(f"Stage: {execution_stats['executionStages']['stage']}")
# 'COLLSCAN' means it searched every single document (slow)
```

```python
db.orders.create_index([("customer_id", 1)])

print("Index created successfully on 'customer_id'!")
```

```python
# Performance WITH Index
print("Performance WITH Index")

execution_stats_indexed = db.orders.find(query_filter).explain()['executionStats']

print(f"Execution Time (ms): {execution_stats_indexed['executionTimeMillis']}ms")
print(f"Total Docs Examined: {execution_stats_indexed['totalDocsExamined']}")
print(f"Stage: {execution_stats_indexed['executionStages']['stage']}")
```
